```python
# we will convert datatypes to appropriate types
# convert release_date to datetime
df['release_date']=pd.to_datetime(df['release_date'],errors='coerce')
```

```python
list_cols=['genres','keywords',"cast"]
for col in list_cols:
    print(f"{col} : {type(df[col].iloc[0])}")
dict_cols=['external_ids']
for col in dict_cols:
    print(f"{col} : {type(df[col].iloc[0])}")
```

```python
# some column are object but should be lists and some should be dictionaries
# we will convert them using eval
list_cols=['genres','keywords',"cast"]
for col in list_cols:
    df[col]=df[col].apply(lambda x: eval(x) if pd.notnull(x) else x)

dict_cols=['external_ids']
for col in dict_cols:
    df[col]=df[col].apply(lambda x: eval(x) if pd.notnull(x) else x)
```

```python
# let's check the datatypes again for confirmation to see if we have lists and dictionaries
list_cols=['genres','keywords',"cast"]
for col in list_cols:
    print(f"{col} : {type(df[col].iloc[0])}")
dict_cols=['external_ids']
for col in dict_cols:
    print(f"{col} : {type(df[col].iloc[0])}")
```

```python
df.dtypes
```

```python
# now our data types are appropriate 
# we got numerics as numeric, dates as datetime, lists as lists, and dictionaries as dictionaries
# python pandas will still show lists and dictionaries as objects but they are now in the correct format as objects could be anything (string, int, list, dict, etc) not number or date
# also we got string columns as object which is correct (overview,title,tagline,etc)
# next we will handle missing values
df.isnull().sum()
```

```python
# overview we will fill na with "No overview available" for tagline we will fill na with ""  so it be blank if no tagline is available
# for release_date since it's datetime we will leave it as and for dispaly purposes we will handle it on the website side making it "Unknown" if nat
# for poster_url we will use "" so on the website we can show a placeholder image if no url is available 
# which is better than na as nan requires special handling on the website side we will need use pd.isna() to check for nan values so using empty string is easier to handle on the website side
# for director we will fill na with "Unknown" for certification we will fill na with "N/A"
df['overview'] = df['overview'].fillna("No overview available")
df['tagline'] = df['tagline'].fillna("")
df['director'] = df['director'].fillna("Unknown")
df['certification'] = df['certification'].fillna("N/A")
df['poster_url'] = df['poster_url'].fillna("")
```
