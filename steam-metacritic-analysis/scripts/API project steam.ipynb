# Steam API

```python
# we will use an API to get steam data
# we will use Steam Spy API to extract most popular games,price,developer etc.

import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup
import json
import time
```

```python
# this retuns data for the first page in jason format
url= 'https://steamspy.com/api.php?request=all&page=0'
response = requests.get(url)
data = response.json()
```

```python
df = pd.DataFrame.from_dict(data, orient='index')
df
```

# now we will request most popular Steam games in 10 pages

```python
from time import time


all_dfs = []
for i in range(10):
    url = f'https://steamspy.com/api.php?request=all&page={i}'
    response = requests.get(url)
    data = response.json()
    df = pd.DataFrame.from_dict(data, orient='index')
    all_dfs.append(df)
    time.sleep(10)  # to avoid overwhelming the server

full_df = pd.concat(all_dfs, ignore_index=True)
```
